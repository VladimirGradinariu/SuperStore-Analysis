```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.head()
```

```python
df.info()
```

```python
df['Month'] = df['Order Date'].dt.to_period('M')
df.head()
```

```python
monthly_revenue = (df.groupby('Month')['Sales'].sum())
monthly_revenue
```

```python
import matplotlib.pyplot as plt
monthly_revenue.plot(kind = 'line', title = 'Monthly Revenue Trend' )
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.show()
```

```python
monthly_profit = (df.groupby('Month')['Profit'].sum())
monthly_profit
```

```python
monthly_profit.plot(kind = 'line', title = 'Monthly Profit Trend' )
plt.xlabel('Month')
plt.ylabel('Profit')
plt.show()
```

```python
plt.figure(figsize=(10,5))

monthly_revenue.plot(label='Monthly Revenue')
monthly_profit.plot(label='Monthly Profit')

plt.title('Monthly Revenue vs Profit')
plt.xlabel('Month')
plt.ylabel('Amount')
plt.legend()
plt.show()
```
